```python
#Intro to Machine Learning
# HW 1

#D3.csv has three explanatory variables x1,x2,x3 and one dependent variable Y

#1) Develop a code that runs linear regression with the gradient descent algorithm
# for each of the explanatory variables in isolation

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

csv_path = 'https://raw.githubusercontent.com/wsteve16/Intro-to-Machine-Learning/main/D3.csv'
df = pd.read_csv(csv_path)

def gradient_descent(x,y,alpha,n):
  theta_0 = 0 # y-intercept
  theta_1 = 0 #slope
  m = len(y)
  cost_tracked = []

  for i in range(n):
    predicted = theta_0 +theta_1*x
    error = predicted-y

    slope_0 = 1/m * sum(error)
    slope_1 = 1/m * sum(error*x)
    theta_0 = theta_0 - alpha*slope_0
    theta_1 = theta_1 - alpha*slope_1

    cost = 1/(2*m) * sum(error**2)
    cost_tracked.append(cost)

  return theta_0,theta_1, cost_tracked


#X1 - alpha of 0.05
x1 = df['X1'].values
y = df['Y'].values

theta_0, theta_1, cost_tracked =gradient_descent(x1, y, 0.05, 500)
print(f"y = {theta_0:.3f} + {theta_1:.3f}*x")
print(f"alpha=0.05: final cost = {cost_tracked[-1]:.3f}")

plt.scatter(x1,y)
x_line = np.linspace(x1.min(),x1.max(),100)
y_line = theta_0 + theta_1*x_line
plt.plot(x_line, y_line)
plt.xlabel('x1')
plt.ylabel('y')
plt.title('Linear Regression: x1 vs y')
plt.legend()
plt.show()

plt.plot(range(len(cost_tracked)), cost_tracked)
plt.xlabel('Iteration')
plt.ylabel('Cost')
plt.title('Loss over Iterations: x1')
plt.show()

#X1- alpha of 0.01
x1 = df['X1'].values
y = df['Y'].values

theta_0, theta_1, cost_tracked =gradient_descent(x1, y, 0.01, 500)
print(f"y = {theta_0:.3f} + {theta_1:.3f}*x")
print(f"alpha=0.01: final cost = {cost_tracked[-1]:.3f}")


plt.scatter(x1,y)
x_line = np.linspace(x1.min(),x1.max(),100)
y_line = theta_0 + theta_1*x_line
plt.plot(x_line, y_line)
plt.xlabel('x1')
plt.ylabel('y')
plt.title('Linear Regression: x1 vs y')
plt.legend()
plt.show()

plt.plot(range(len(cost_tracked)), cost_tracked)
plt.xlabel('Iteration')
plt.ylabel('Cost')
plt.title('Loss over Iterations: x1')
plt.show()

#X1 - alpha of 0.1
x1 = df['X1'].values
y = df['Y'].values

theta_0, theta_1, cost_tracked =gradient_descent(x1, y, 0.1, 500)
print(f"y = {theta_0:.3f} + {theta_1:.3f}*x")
print(f"alpha=0.1: final cost = {cost_tracked[-1]:.3f}")


plt.scatter(x1,y)
x_line = np.linspace(x1.min(),x1.max(),100)
y_line = theta_0 + theta_1*x_line
plt.plot(x_line, y_line)
plt.xlabel('x1')
plt.ylabel('y')
plt.title('Linear Regression: x1 vs y')
plt.legend()
plt.show()

plt.plot(range(len(cost_tracked)), cost_tracked)
plt.xlabel('Iteration')
plt.ylabel('Cost')
plt.title('Loss over Iterations: x1')
plt.show()

#X2
x2 = df['X2'].values
y = df['Y'].values

theta_0_x2, theta_1_x2, cost_tracked_x2 = gradient_descent(x2, y, 0.05, 500)
print(f"y = {theta_0_x2:.3f} + {theta_1_x2:.3f}*x")
print(f"cost: {cost_tracked_x2[-1]:.3f}")

plt.scatter(x2, y)
x_line = np.linspace(x2.min(), x2.max(), 100)
y_line = theta_0_x2 + theta_1_x2*x_line
plt.plot(x_line, y_line)
plt.xlabel('x2')
plt.ylabel('y')
plt.title('Linear Regression: x2 vs y')
plt.legend()
plt.show()

plt.plot(range(len(cost_tracked_x2)), cost_tracked_x2)
plt.xlabel('Iteration')
plt.ylabel('Cost')
plt.title('Loss over Iterations: x2')
plt.show()

#X3
x3 = df['X3'].values
y = df['Y'].values

theta_0_x3, theta_1_x3, cost_tracked_x3 = gradient_descent(x3, y, 0.05, 500)
print(f"y = {theta_0_x3:.3f} + {theta_1_x3:.3f}*x")
print(f"cost: {cost_tracked_x3[-1]:.3f}")

plt.scatter(x3, y)
x_line = np.linspace(x3.min(), x3.max(), 100)
y_line = theta_0_x3 + theta_1_x3*x_line
plt.plot(x_line, y_line)
plt.xlabel('x3')
plt.ylabel('y')
plt.title('Linear Regression: x3 vs y')
plt.legend()
plt.show()

plt.plot(range(len(cost_tracked_x3)), cost_tracked_x3)
plt.xlabel('Iteration')
plt.ylabel('Cost')
plt.title('Loss over Iterations: x3')
plt.show()


#Question 2- linear regression with gradient descent algorithm using all three explanatory variables
def gradient_descent_multiple(x1,x2,x3,y,alpha,n):
  theta_0 = 0
  theta_1 = 0
  theta_2 = 0
  theta_3 = 0
  m = len(y)
  cost_tracked = []

  for i in range(n):
    predicted = theta_0 + theta_1*x1 + theta_2*x2 + theta_3*x3
    error = predicted - y

    slope_0 = 1/m * sum(error)
    slope_1 = 1/m * sum(error*x1)
    slope_2 = 1/m * sum(error*x2)
    slope_3 = 1/m * sum(error*x3)

    theta_0 = theta_0 - alpha*slope_0
    theta_1 = theta_1 - alpha*slope_1
    theta_2 = theta_2 - alpha*slope_2
    theta_3 = theta_3 - alpha*slope_3

    cost = 1/(2*m) * sum(error**2)
    cost_tracked.append(cost)

  return theta_0, theta_1, theta_2, theta_3, cost_tracked


theta_0, theta_1, theta_2, theta_3, cost_tracked = gradient_descent_multiple(x1, x2, x3, y, 0.05, 500)

print(f"y = {theta_0:.3f} + {theta_1:.3f}*x1 + {theta_2:.3f}*x2 + {theta_3:.3f}*x3")
print(f"final cost: {cost_tracked[-1]:.4f}")

#loss over iterations plot
plt.plot(range(len(cost_tracked)), cost_tracked)
plt.xlabel('Iteration')
plt.ylabel('Cost')
plt.title('Loss over Iterations for Multiple Variables')
plt.show()

#compare alphas
#alpha of 0.05
theta_0, theta_1, theta_2, theta_3, cost_tracked = gradient_descent_multiple(x1, x2, x3, y, 0.05, 500)
print(f"final cost of alpha 0.05: {cost_tracked[-1]:.4f}")
#alpha of 0.01
theta_0, theta_1, theta_2, theta_3, cost_tracked = gradient_descent_multiple(x1, x2, x3, y, 0.01, 500)
print(f"final cost of alpha 0.01: {cost_tracked[-1]:.4f}")
#alpha of 0.1
theta_0, theta_1, theta_2, theta_3, cost_tracked = gradient_descent_multiple(x1, x2, x3, y, 0.1, 500)
print(f"final cost of alpha 0.1: {cost_tracked[-1]:.4f}")


#predict values with alpha of 0.1 because that had lowest cost
# (1,1,1)
point_1 = theta_0 +theta_1*1 +theta_2*1 +theta_3*1
print(f"Value of y at (1,1,1): {point_1:.3f}")

# (2,0,4)
point_2 = theta_0 +theta_1*2 +theta_2*0 +theta_3*4
print(f"Value of y at (2,0,4): {point_2:.3f}")

# (3,2,1)
point_3 = theta_0 +theta_1*3 +theta_2*2 +theta_3*1
print(f"Value of y at (3,2,1): {point_3:.3f}")


```
